# 1

1. Let $\ell = -\log(\sigma) - \frac{1}{2\sigma^2} \sum_{i=1}^n(y_i - x_i \beta)^2$. Find the value of $\mu$ that maximizes $\ell$ by setting $\ell'(\mu) = 0$ and solving. Check that the second derivative of $\ell(\mu)$ is negative at that value of $\mu$; why do we care? Find the value of $\sigma$ that maximizes $\ell$ by setting $\ell'(\sigma)=0$ and solving.

The expression given is

$$
\ell
= -\log(\sigma)
-\frac{1}{2\sigma^2}\sum_{i=1}^n(y_i-x_i\beta)^2.
$$

There is no $\mu$ in this expression, and no relationship between $\mu$ and $\beta$ is specified. Treating $\beta$ and $\sigma$ as fixed,

$$
\frac{\partial\ell}{\partial\mu}=0,
\qquad
\frac{\partial^2\ell}{\partial\mu^2}=0.
$$

Thus, setting the first derivative equal to zero does not identify a particular value of $\mu$, and the second derivative is not negative. A negative second derivative at a stationary point would establish a strict local maximum, but that check cannot be satisfied for $\mu$ in the expression as written.

To maximize over $\sigma$ for fixed $\beta$, let

$$
S=\sum_{i=1}^n(y_i-x_i\beta)^2.
$$

Then

$$
\ell(\sigma)=-\log(\sigma)-\frac{S}{2\sigma^2}.
$$

Differentiating and setting the derivative equal to zero gives

$$
\begin{aligned}
\ell'(\sigma)
&= -\frac{1}{\sigma}+\frac{S}{\sigma^3},\\
0 &= -\hat{\sigma}^2+S,\\
\hat{\sigma}^2 &= S.
\end{aligned}
$$

Therefore, when S > 0,

$$
\hat{\sigma}
= \sqrt{\sum_{i=1}^n(y_i-x_i\beta)^2}.
$$

The second derivative at this solution is

$$
\ell''(\hat{\sigma})
= \frac{1}{\hat{\sigma}^2}
-\frac{3S}{\hat{\sigma}^4}
= -\frac{2}{\hat{\sigma}^2}<0.
$$

Also, $\ell'(\sigma)=(S-\sigma^2)/\sigma^3$ is positive below $\sqrt{S}$ and negative above it, so this is the maximizing value.

If S = 0, then $\ell(\sigma)=-\log(\sigma)$ increases without bound as $\sigma$ approaches zero, so no finite maximizer exists for $\sigma$ > 0.

## 3.A

Since $y\sim N(X\beta,\sigma^2I)$, the multivariate normal likelihood is

$$
\mathcal{L}(\beta,\sigma^2)
= \frac{1}{(2\pi)^{n/2}|\sigma^2I|^{1/2}}
\exp\left\{
-\frac{1}{2}(y-X\beta)^\top(\sigma^2I)^{-1}(y-X\beta)
\right\}.
$$

The covariance matrix has determinant and inverse

$$
|\sigma^2I|=(\sigma^2)^n,
\qquad
(\sigma^2I)^{-1}=\frac{1}{\sigma^2}I.
$$

Substituting gives

$$
\mathcal{L}(\beta,\sigma^2)
= (2\pi\sigma^2)^{-n/2}
\exp\left\{
-\frac{1}{2\sigma^2}(y-X\beta)^\top(y-X\beta)
\right\}.
$$

Taking the logarithm gives

$$
\begin{aligned}
\ell(\beta,\sigma^2)
&= -\frac{n}{2}\log(2\pi\sigma^2)
-\frac{1}{2\sigma^2}(y-X\beta)^\top(y-X\beta)\\
&= -\frac{n}{2}\log(2\pi)
-\frac{n}{2}\log(\sigma^2)
-\frac{1}{2\sigma^2}(y-X\beta)^\top(y-X\beta).
\end{aligned}
$$

## 3.B

For fixed $\sigma^2$ > 0, the first two terms of the log-likelihood do not depend on $\beta$. The remaining term is a negative constant times the sum of squared errors. Therefore, maximizing $\ell$ over $\beta$ is equivalent to minimizing

$$
S(\beta)=(y-X\beta)^\top(y-X\beta).
$$

Expanding gives

$$
S(\beta)
= y^\top y-2\beta^\top X^\top y+\beta^\top X^\top X\beta.
$$

Taking the gradient and setting it equal to zero gives

$$
\begin{aligned}
\nabla_\beta S(\beta)
&= -2X^\top y+2X^\top X\beta,\\
0 &= -2X^\top y+2X^\top X\hat{\beta},\\
X^\top X\hat{\beta} &= X^\top y.
\end{aligned}
$$

Assuming X has full column rank, $X^\top$ X is invertible, so

$$
\hat{\beta}=(X^\top X)^{-1}X^\top y.
$$

The Hessian of S($\beta$) is 2 $X^\top$ X, which is positive definite under this assumption. Thus, $\hat{\beta}$ uniquely minimizes the sum of squared errors and maximizes the log-likelihood over $\beta$.
